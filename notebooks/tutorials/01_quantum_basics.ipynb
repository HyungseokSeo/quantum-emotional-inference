## 2. Ambivalence vs Uncertainty

The key distinction in this framework.

```python
# Ambivalent state: genuinely feeling both happy and sad
rho_ambivalent = basis.ambivalent(["happy", "sad"], [0.5, 0.5])

# Uncertain state: don't know if happy or sad
rho_uncertain = basis.uncertain(["happy", "sad"], [0.5, 0.5])

print(f"Ambivalent coherence: {rho_ambivalent.l1_coherence():.4f}")
print(f"Uncertain coherence: {rho_uncertain.l1_coherence():.4f}")
```
